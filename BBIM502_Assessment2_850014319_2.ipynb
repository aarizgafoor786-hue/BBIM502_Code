```python
# File Name: BBIM502_Assessment2_850014319
# Author: Aariz Gafoor Ali
# Description: Langham Hotel Management System for managing rooms, customer allocations, billing, and file handling.
# Date: 01/11/2026

# Importing the required modules
try:
    import datetime
    import os
except ImportError:
    # Catching ImportError when a required module cannot be imported
    print("Error: Required module could not be imported.")
    raise

# Creating empty lists to store rooms and room allocations
list_of_rooms = []
list_of_room_allocations = []


# 1. Function for adding rooms
def add_rooms():
    #This function allows me to add new rooms to the hotel.
    try:
        count = int(input("How many rooms do you want to add: "))

        # FIX: Checking that the room count is greater than zero
        if count <= 0:
            print("Please enter a number greater than zero.")
            return

        for i in range(count):
            print("\nEntering details for Room", i + 1)
            room_no = int(input("Enter Room Number: "))

            # FIX: Checking whether the room number already exists
            duplicate = False
            for room in list_of_rooms:
                if room["room_no"] == room_no:
                    duplicate = True
                    break

            if duplicate:
                print("This room number already exists.")
            else:
                room = {"room_no": room_no, "is_allocated": False}
                list_of_rooms.append(room)
                print("Room added successfully.")

        print("\nRoom operation completed.")

    except ValueError:
        # Catching ValueError when text or an invalid number is entered
        print("Error: Please enter valid numeric values.")
    except TypeError:
        # Catching TypeError when an operation receives an incorrect data type
        print("Error: Incorrect data type entered.")
    except EOFError:
        # Catching EOFError when input ends unexpectedly
        print("Error: No input received.")


# 2. Function for deleting rooms
def delete_rooms():
#This function allows me to delete rooms that are not allocated.
    if len(list_of_rooms) == 0:
        print("No rooms have been added yet.")
        return

    try:
        room_no = int(input("Enter the room number to delete: "))
        room_found = False

        for room in list_of_rooms:
            if room["room_no"] == room_no:
                room_found = True

                # FIX: Preventing deletion of a room that is allocated
                if room["is_allocated"]:
                    print("Cannot delete an allocated room.")
                else:
                    list_of_rooms.remove(room)
                    print("Room deleted successfully.")
                break

        if not room_found:
            print("Room number not found.")

    except ValueError:
        # Catching ValueError when the room number is not numeric
        print("Error: Please enter a numeric room number.")
    except EOFError:
        # Catching EOFError when input ends unexpectedly
        print("Error: No input received.")


# 3. Function for displaying room details
def display_rooms():
    #This function displays all rooms and their current status.
    if len(list_of_rooms) == 0:
        print("No rooms have been added yet.")
        return

    print("\n----- LIST OF ROOMS -----")
    try:
        # Displaying rooms by list index
        for index in range(len(list_of_rooms)):
            room = list_of_rooms[index]
            print("Room Number:", room["room_no"])
            print("Room Status:", "Allocated" if room["is_allocated"] else "Available")
            print("-------------------------")

    except IndexError:
        # Catching IndexError if a list position is no longer available
        print("Error: Room list position was not found.")
    except TypeError:
        # Catching TypeError if room information has an incorrect type
        print("Error: Incorrect room data type.")


# 4. Function for allocating rooms
def allocate_rooms():
    #This function allows me to allocate an available room to a customer.
    if len(list_of_rooms) == 0:
        print("No rooms have been added yet.")
        return

    try:
        room_no = int(input("Enter the room number to allocate: "))
        room_found = False

        for room in list_of_rooms:
            if room["room_no"] == room_no:
                room_found = True

                # FIX: Preventing an already allocated room from being booked again
                if room["is_allocated"]:
                    print("The room is already allocated.")
                    return

                customer_no = int(input("Enter Customer Number: "))
                customer_name = input("Enter Customer Name: ")
                nights = int(input("Enter Number of Nights: "))

                # FIX: Rejecting zero or negative nights
                if nights <= 0:
                    print("Number of nights must be greater than zero.")
                    return

                # FIX: Checking whether this customer number is already in use
                duplicate = False
                for allocation in list_of_room_allocations:
                    if allocation["customer_no"] == customer_no:
                        duplicate = True
                        break

                if duplicate:
                    print("Customer number already exists.")
                    return

                allocation = {
                    "room_no": room_no,
                    "customer_no": customer_no,
                    "customer_name": customer_name,
                    "nights": nights,
                    "date": datetime.datetime.now()
                }

                list_of_room_allocations.append(allocation)
                room["is_allocated"] = True
                print("Room allocated successfully to", customer_name)
                break

        # FIX: Informing the user if the requested room number does not exist
        if not room_found:
            print("Room number does not exist.")

    except ValueError:
        # Catching ValueError when numeric input is invalid
        print("Error: Please enter valid numeric values.")
    except TypeError:
        # Catching TypeError when data types are incompatible
        print("Error: Incorrect data type entered.")
    except EOFError:
        # Catching EOFError when input ends unexpectedly
        print("Error: No input received.")


# 5. Function for displaying room allocation details
def display_room_allocations():
    #This function displays details of currently allocated rooms.
    if len(list_of_room_allocations) == 0:
        print("No rooms are currently allocated.")
        return

    print("\n----- ROOM ALLOCATION DETAILS -----")
    for allocation in list_of_room_allocations:
        print("Room Number:", allocation["room_no"])
        print("Customer Number:", allocation["customer_no"])
        print("Customer Name:", allocation["customer_name"])
        print("Number of Nights:", allocation["nights"])
        print("Allocation Date:", allocation["date"])
        print("-----------------------------------")


# 6. Function for billing and de-allocation
def billing_and_deallocation():
    #This function calculates the bill and releases the room. Price is $100 per night.
    if len(list_of_room_allocations) == 0:
        print("No room allocations exist.")
        return

    try:
        room_no = int(input("Enter room number for billing: "))
        allocation_found = False

        for allocation in list_of_room_allocations:
            if allocation["room_no"] == room_no:
                allocation_found = True
                price_per_night = 100
                nights = allocation["nights"]

                # FIX: Handling an unexpected zero-night allocation before division
                try:
                    total_bill = price_per_night * nights
                    average_rate = total_bill / nights
                except ZeroDivisionError:
                    # Catching ZeroDivisionError if nights is zero
                    print("Error: Cannot calculate a bill for zero nights.")
                    return

                # Catching OverflowError if a very large calculation exceeds float limits
                try:
                    checked_total = float(total_bill)
                except OverflowError:
                    print("Error: The bill amount is too large to calculate.")
                    return

                print("\n========== HOTEL BILL ==========")
                print("Customer Number:", allocation["customer_no"])
                print("Customer Name:", allocation["customer_name"])
                print("Room Number:", allocation["room_no"])
                print("Check-in Date:", allocation["date"])
                checkout_date = datetime.datetime.now()
                print("Check-out Date:", checkout_date)
                print("Number of Nights:", nights)
                print("Price Per Night: $", price_per_night)
                print("Total Bill: $", checked_total)
                print("Average Rate: $", average_rate)
                print("================================")

                list_of_room_allocations.remove(allocation)

                
                for room in list_of_rooms:
                    if room["room_no"] == room_no:
                        room["is_allocated"] = False
                        break

                print("Room successfully de-allocated.")
                break

        # FIX: Preventing billing when there is no allocation for that room
        if not allocation_found:
            print("No allocation found for this room.")

    except ValueError:
        # Catching ValueError when the entered room number is not numeric
        print("Error: Please enter a numeric room number.")
    except TypeError:
        # Catching TypeError when billing data has an incompatible type
        print("Error: Incorrect data type entered.")
    except EOFError:
        # Catching EOFError when input ends unexpectedly
        print("Error: No input received.")


# 7. Function for saving room allocations to a file
def save_room_allocations():
    #This function saves room allocation details to a text file.
    filename = "LHMS_850014319.txt"

    try:
        file = open(filename, "w")
        for allocation in list_of_room_allocations:
            file.write("Room Number: " + str(allocation["room_no"]) + "\n")
            file.write("Customer Number: " + str(allocation["customer_no"]) + "\n")
            file.write("Customer Name: " + allocation["customer_name"] + "\n")
            file.write("Number of Nights: " + str(allocation["nights"]) + "\n")
            file.write("Allocation Date: " + str(allocation["date"]) + "\n")
            file.write("-----------------------------------\n")
        file.close()
        print("Room allocations saved to file successfully.")

    except IOError:
        # Catching IOError when a file cannot be opened, written, or closed
        print("Error: Unable to perform file input/output operation.")
    except TypeError:
        # Catching TypeError if file content has an incompatible data type
        print("Error: Incorrect data type when writing to file.")


# 8. Function for displaying room allocations from a file
def display_room_allocations_from_file():
    #This function reads allocation details from a text file.
    filename = "LHMS_850014319.txt"

    try:
        file = open(filename, "r")
        content = file.read()
        print("\n----- ROOM ALLOCATIONS FROM FILE -----")
        print(content)
        file.close()

    except FileNotFoundError:
        # Catching FileNotFoundError when the requested file does not exist
        print("Error: File not found. Please save allocations first.")
    except IOError:
        # Catching IOError for other file reading problems
        print("Error: Unable to read the file.")
    except TypeError:
        # Catching TypeError if file content is not the expected type
        print("Error: Incorrect data type when reading the file.")


# 9. Function for backing up and clearing room allocation file
def backup_and_clear_room_allocations():
    #This function backs up the allocation file and clears the original.
    filename = "LHMS_850014319.txt"
    current_datetime = datetime.datetime.now()
    backup_filename = (
        "LHMS_850014319_Backup_"
        + current_datetime.strftime("%Y-%m-%d_%H-%M-%S")
        + ".txt"
    )

    try:
        file = open(filename, "r")
        content = file.read()
        file.close()

        backup_file = open(backup_filename, "a")
        backup_file.write(content)
        backup_file.close()

        file = open(filename, "w")
        file.close()

        print("Backup created successfully.")
        print("Room allocation file cleared successfully.")

    except FileNotFoundError:
        # Catching FileNotFoundError when the original allocation file is missing
        print("Error: Allocation file not found. Save allocations first.")
    except IOError:
        # Catching IOError during file backup, reading, writing, or clearing
        print("Error: Unable to perform file operation.")
    except TypeError:
        # Catching TypeError if file data is an incorrect type
        print("Error: Incorrect data type during file operation.")


# 10. Function for exiting the application
def exit_application():
    #This function displays a message when the user exits.
    print("Exiting the application. Goodbye!")


# Main function for displaying the menu
def main():
    #This function displays the menu and calls the selected option.
    while True:
        print("\n********************************************")
        print("       LANGHAM HOTEL MANAGEMENT SYSTEM")
        print("********************************************")
        print("1. Add Rooms")
        print("2. Delete Rooms")
        print("3. Display Room Details")
        print("4. Allocate Rooms")
        print("5. Display Room Allocation Details")
        print("6. Billing and De-Allocation")
        print("7. Save Room Allocation to File")
        print("8. Display Room Allocation from File")
        print("9. Backup and Clear Room Allocation File")
        print("0. Exit Application")
        print("********************************************")

        try:
            choice = input("Enter Your Choice Number Here: ")
        # Catching EOFError when menu input ends unexpectedly
        except EOFError:
            print("Error: Input ended unexpectedly. Exiting application.")
            break

        try:
            if choice == "1":
                add_rooms()
            elif choice == "2":
                delete_rooms()
            elif choice == "3":
                display_rooms()
            elif choice == "4":
                allocate_rooms()
            elif choice == "5":
                display_room_allocations()
            elif choice == "6":
                billing_and_deallocation()
            elif choice == "7":
                save_room_allocations()
            elif choice == "8":
                display_room_allocations_from_file()
            elif choice == "9":
                backup_and_clear_room_allocations()
            elif choice == "0":
                exit_application()
                break
            else:
                print("Invalid choice! Please select a number from 0 to 9.")

        # Catching NameError if a variable name is accidentally undefined during execution
        except NameError:
            print("Error: A variable name was not defined correctly.")
        # Catching SyntaxError from dynamically compiled Python code
        except SyntaxError:
            print("Error: A syntax problem was detected.")
        # Catching TypeError for incompatible types passed during menu operations
        except TypeError:
            print("Error: An operation used an incorrect data type.")


# Calling the main function to start the program
if __name__ == "__main__":
    main()
```
